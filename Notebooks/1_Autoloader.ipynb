```python
%sql
CREATE SCHEMA netflix_catalog.net_schema;
```

```python
df = spark.readStream\
    .format("cloudFiles") \
    .option("cloudFiles.format", "csv") \
    .option("cloudFiles.schemaLocation", "abfss://silver@netflixdatalake.dfs.core.windows.net/CheckpointLocation")\
    .load("abfss://raw@netflixdatalake.dfs.core.windows.net")
```

```python
df.writeStream \
    .format("delta") \
    .option("checkpointLocation","abfss://silver@netflixdatalake.dfs.core.windows.net/CheckpointLocation") \
    .outputMode("append") \
    .start("abfss://bronze@netflixdatalake.dfs.core.windows.net/netflix_titles")
```
